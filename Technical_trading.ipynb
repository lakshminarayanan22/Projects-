```python
import yfinance as yf
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# Define the stock lists
bank_stock_list = ["HDFCBANK.NS", "SBIN.NS", "ICICIBANK.NS", "BANKBARODA.NS", "UNIONBANK.NS", "INDUSINDBK.NS", "AXISBANK.NS"]
med_stock_list = ["CIPLA.NS", "LAURUSLABS.NS", "AUROPHARMA.NS", "SUNPHARMA.NS", "ZYDUSLIFE.NS", "ABBOTINDIA.NS"]

# List to hold the bank stock data
bank_stock_data = []
med_stock_data = []
it_stock_data = []

for ticker in bank_stock_list:
    stock = yf.Ticker(ticker)
    data = stock.history(period="5y")
    df = pd.DataFrame(data)
    df = df.drop(["High", "Low", "Open", "Volume"], axis=1)
    std_dev = df['Close'].std()
    bank_stock_data.append([ticker, df,std_dev])

for ticker in med_stock_list:
    stock = yf.Ticker(ticker)
    data = stock.history(period="5y")
    df = pd.DataFrame(data)
    df = df.drop(["High", "Low", "Open", "Volume"], axis=1)
    std_dev = df['Close'].std()
    med_stock_data.append([ticker, df,std_dev])


def thirty_day_sma(dataframe):
    dataframe['30_DAY_SMA'] = dataframe['Close'].rolling(window=30).mean()

def sixty_day_sma(dataframe):
    dataframe['60_DAY_SMA'] = dataframe['Close'].rolling(window = 60).mean()

def sma_200(dataframe):
    dataframe['200_day_sma'] = dataframe['Close'].rolling(window = 200).mean()

def rsi(dataframe):
    rsi = 0
    gain,loss = 0,0
    gaincount,losscount = 0,0
    for i in range(dataframe.shape[0]-1):
        if dataframe.iloc[i]["Close"]<=dataframe.iloc[i+1]["Close"]:
            gain += dataframe.iloc[i+1]["Close"] - dataframe.iloc[i]["Close"]
            gaincount += 1
        else:
            loss += dataframe.iloc[i]["Close"] - dataframe.iloc[i+1]["Close"]
            losscount += 1
    avg_gain = gain/gaincount
    avg_loss = loss/losscount
    rsi = 100 - (100/(1+(avg_gain/avg_loss)))
    return rsi

def thirty_day_moving_std_dev(dataframe):
    return dataframe['30_DAY_SMA'].rolling(window=30).std()

def sixty_day_moving_std_dev(dataframe):
    return dataframe['60_DAY_SMA'].rolling(window = 60).std()

def standard_dev(dataframe):
    sigma  = dataframe['Close'].std()
    return sigma

def mean(dataframe):
    mean = dataframe['Close'].mean()
    return mean

def bollinger_band_30_day(dataframe):
    dataframe['30_day_upper_bollinger_band'] = dataframe['Close'].rolling(window=30).mean()+(2*(thirty_day_moving_std_dev(dataframe)))
    dataframe['30_day_lower_bollinger_band'] = dataframe['Close'].rolling(window=30).mean()-(2*(thirty_day_moving_std_dev(dataframe)))

def macd(dataframe):
    dataframe['MACD'] = (dataframe['Close'].ewm(span=12, adjust=False).mean()) - (dataframe['Close'].ewm(span=26, adjust=False).mean())
    dataframe['MACD_Signal_line'] = dataframe['MACD'].ewm(span=9, adjust=False).mean()

def plot(dataframe):
    length = dataframe.shape[0]
    x = []
    y = []
    y_30 = []
    y_60 = []
    y_200 = []
    y_30_u = []
    y_30_l = []
    y_macd = []
    y_macd_signal = []
    for j in range(length):
        x.append(j)
        y.append(dataframe.iloc[j]["Close"])
        y_30.append(dataframe.iloc[j]['30_DAY_SMA'])
        y_60.append(dataframe.iloc[j]['60_DAY_SMA'])
        y_200.append(dataframe.iloc[j]['200_day_sma'])
        y_30_u.append(dataframe.iloc[j]['30_day_upper_bollinger_band'])
        y_30_l.append(dataframe.iloc[j]['30_day_lower_bollinger_band'])
        y_macd.append(dataframe.iloc[j]["MACD"])
        y_macd_signal.append(dataframe.iloc[j]["MACD_Signal_line"])

    plt.figure(figsize=(8, 5))
    plt.plot(x,y,label = 'closing prices')
    plt.plot(x,y_30,label = '30 day sma')
    plt.plot(x,y_60,label = '60 day sma')
    plt.plot(x,y_200,label = '200 day sma')
    plt.grid(True)
    plt.legend()
    plt.show()
    plt.plot(x,y,label = 'closing prices')
    plt.tight_layout()
    plt.plot(x,y_30_u,label = 'upper_bollinger_band')
    plt.plot(x,y_30_l,label = 'lower_bollinger_band')
    plt.grid(True)
    plt.legend()
    plt.show()
    plt.plot(x,y,label = 'Closing prices')
    plt.plot(x,y_macd,label = "MACD")
    plt.plot(x,y_macd_signal,label = "signal line")
    plt.grid(True)
    plt.legend()
    plt.show()

for stock in med_stock_data:
    thirty_day_sma(stock[1])
    sixty_day_sma(stock[1])
    sma_200(stock[1])
    bollinger_band_30_day(stock[1])
    macd(stock[1])
    print(stock[0],"\n")
    print("mean:",mean(stock[1]),"\n")
    print("standard deviation:",standard_dev(stock[1]),"\n")
    print(stock[1],"\n")
    plot(stock[1])
```

```python
#RL environment framework

#med_stock_list = ["CIPLA.NS", "LAURUSLABS.NS", "AUROPHARMA.NS", "SUNPHARMA.NS", "ZYDUSLIFE.NS", "ABBOTINDIA.NS"]

closing_price_list = [[],[],[],[],[],[]]
ma_30_list = [[],[],[],[],[],[]]
ma_60_list = [[],[],[],[],[],[]]
ma_200_list = [[],[],[],[],[],[]]
u_bb_list = [[],[],[],[],[],[]]
l_bb_list = [[],[],[],[],[],[]]
macd_list = [[],[],[],[],[],[]]
signal_list = [[],[],[],[],[],[]]
for j in range(len(med_stock_data)):
    df = med_stock_data[j][1]
    for i in range(df.shape[0]):
        closing_price_list[j].append(df.iloc[i]['Close'])
        ma_30_list[j].append(df.iloc[i]['30_DAY_SMA'])
        ma_60_list[j].append(df.iloc[i]['60_DAY_SMA'])
        ma_200_list[j].append(df.iloc[i]['200_day_sma'])
        u_bb_list[j].append(df.iloc[i]['30_day_upper_bollinger_band'])
        l_bb_list[j].append(df.iloc[i]['30_day_lower_bollinger_band'])
        macd_list[j].append(df.iloc[i]['MACD'])
        signal_list[j].append(df.iloc[i]['MACD_Signal_line'])

print(closing_price_list)

def give_plain_state_t(t):
    plain_state = []
    price_list = []
    ma_30_l = []
    ma_60_l = []
    ma_200_l = []
    u_bb_l = []
    l_bb_l = []
    macd_l = []
    signal_l = []
    #We take the values from day 200 because aftter that point the data will include the 200_day_moving average and every other metric.
    for i in range(len(med_stock_data)):
        price_list.append(closing_price_list[i][t])
        ma_30_l.append(ma_30_list[i][t])
        ma_60_l.append(ma_60_list[i][t])
        ma_200_l.append(ma_200_list[i][t])
        u_bb_l.append(u_bb_list[i][t])
        l_bb_l.append(l_bb_list[i][t])
        macd_l.append(macd_list[i][t])
        signal_l.append(signal_list[i][t])
    plain_state.append(price_list)
    plain_state.append(ma_30_l)
    plain_state.append(ma_60_l)
    plain_state.append(ma_200_l)
    plain_state.append(u_bb_l)
    plain_state.append(l_bb_l)
    plain_state.append(macd_l)
    plain_state.append(signal_l)
    return plain_state
```

```python

#moving average policy
plain_state_list = []

for k in range(200, 1200):
    plain_state = give_plain_state_t(k)
    plain_state_list.append(plain_state)

n_list = [[0, 0, 0, 0, 0, 0]]  # Initial holdings
cash_list = [100000]  # Initial cash

def ma_policy(t,l):
    if t == 200:
        # Initial state
        cash = cash_list[0]
        n = [0] * l
        if not plain_state_list:  # Ensure plain_state_list is not empty
            raise ValueError("plain_state_list is empty. Ensure it's populated correctly.")
        x = plain_state_list[0]
        x.insert(0, n)
        return [cash, x]
    else:
        if t - 200 >= len(plain_state_list) or t - 200 - 1 >= len(plain_state_list):
            raise IndexError(f"Index {t - 200} or {t - 200 - 1} is out of range for plain_state_list.")

        x0 = plain_state_list[t - 200 - 1].copy()
        x = plain_state_list[t - 200].copy()

        n_prev = n_list[t - 200 - 1]
        n = n_prev.copy()

        for j in range(l):
            if (x0[1][j] < x0[3][j]) and (x[1][j] >= x[3][j]) and cash_list[t - 200 - 1] > x[0][j]:
                n[j] += (cash_list[t-200-1]/4)//(x[0][j])
            elif (x0[2][j] < x0[3][j]) and (x[2][j] >= x[3][j]) and cash_list[t - 200 - 1] > x[0][j]:
                n[j] += (cash_list[t-200-1]/4)//(x[0][j])
            elif (x0[1][j] >= x0[3][j]) and (x[1][j] < x[3][j]) and n[j] > 0:
                n[j] = 0
            elif (x0[2][j] >= x0[3][j]) and (x[2][j] < x[3][j]) and n[j] > 0:
                n[j] = 0

    n_list.append(n)
    x.insert(0, n)

    # Update cash balance
    exchange_cash = 0
    for k in range(l):
        delta_n = n[k] - n_prev[k]
        exchange_cash -= delta_n *(x[1][k])

    cash = cash_list[t - 200 - 1] + exchange_cash
    cash_list.append(cash)

    return [cash, x]

def reward(final_state,l):
    score = final_state[0]
    for j in range(l):
        score += final_state[1][0][j] * final_state[1][1][j]
    return score


state_list = []
for k in range(200, 1200):  # Adjusted range to match plain_state_list
        m = ma_policy(k,6)
        state_list.append(m)
        print(m)
        print("Reward:", reward(m,6))

```

```python
import numpy as np

def greedy_policy(t, l):
    def n_max_possible(t, l):
        prices = np.array(price_list[t - 200])
        cash = cash_list[-1]
        return np.floor(cash / prices).astype(int)

    if t == 200:
        cash = cash_list[0]
        n = np.zeros(l, dtype=int)
        if not price_list:
            raise ValueError("price_list is empty. Ensure it's populated correctly.")
        x = np.array(price_list[0])
        x = np.insert(x, 0, n)
        return [cash, x]
    else:
        x0 = np.array(price_list[t - 200 - 1])
        x = np.array(price_list[t - 200])
        n_prev = np.array(n_list[t - 200 - 1])
        n = n_prev.copy()
        n_max = n_max_possible(t, l)

        cash = cash_list[t - 200 - 1]
        best_n = n.copy()
        best_cash_exchange = 0

        buy_n = np.minimum(n_max, (cash // x).astype(int)) * 0.2
        buy_n = buy_n.astype(int)
        total_buy_cost = np.sum(buy_n * x)

        sell_n = (n_prev * 0.2).astype(int)
        sell_cash = np.sum(sell_n * x)

        for k in range(l):
            if total_buy_cost <= cash:
                possible_n = np.arange(n_max[k] + 1) * 0.2
                possible_n = possible_n.astype(int)

                exchange_cash = -np.sum((possible_n - n_prev[k]) * x[k])

                if -exchange_cash <= cash:
                    if -exchange_cash > best_cash_exchange:
                        best_cash_exchange = exchange_cash
                        best_n[k] = possible_n[np.argmax(exchange_cash)]

        best_action = np.argmax([cash, cash + sell_cash, cash - total_buy_cost])

        if best_action == 1:
            cash += sell_cash
            best_n -= sell_n
        elif best_action == 2 and total_buy_cost <= cash:
            cash -= total_buy_cost
            best_n += buy_n

        cash_list.append(cash)
        n_list.append(best_n.tolist())
        x = np.insert(x, 0, best_n)

    return [cash, x]

def reward(final_state, l):
    cash = final_state[0]
    value_of_holdings = np.sum(final_state[1][:l] * final_state[1][l:])  # Compute portfolio value
    return cash + value_of_holdings

state_list = []
for k in range(200, 1200):
    m = greedy_policy(k, 6)
    state_list.append(m)
    print(m)
    print("Reward:", reward(m, 6))
```
